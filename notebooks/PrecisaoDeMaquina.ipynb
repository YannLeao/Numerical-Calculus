Aqui vou explora o conceito de **precisão de máquina** (ou epsilon da máquina), que representa o menor número positivo tal que `1 + ε > 1` em ponto flutuante.

### 1. Algoritmo para epsilon da máquina (float padrão do Python)

O algoritmo abaixo encontra o epsilon da máquina, ou seja, o menor valor que limita a igualdade `(a + b) == a`, ou seja, o **epsilon** é tão pequeno que não altera mais o número

```python
epsilon = 1.0
soma = 1.0 + epsilon
iteracoes = 0

while soma > 1.0:
    epsilon /= 2.0
    soma = 1.0 + epsilon
    iteracoes += 1

print(f"Precisão de máquina: {epsilon * 2}")
print(f"Iterações até parar: {iteracoes}")
```

**Note que, ao final, devemos multiplicar o epsilon por 2, já que a ideia é retornar o último valor que ainda afetava a soma.**

## Comparação de precisões com diferentes tipos de ponto flutuante através do `numpy.finfo`

A biblioteca `numpy` oferece o método `finfo()`, que retorna informações sobre tipos de ponto flutuante como:
- `float16` (16 bits)
- `float32` (32 bits)
- `float64` (64 bits)

> Por padrão o Python utiliza float64


```python
import numpy as np
import pandas as pd # A biblioteca Pandas foi utilizada apenas para uma melhor vizualização

tipos = [np.float16, np.float32, np.float64]
dados = []

for tipo in tipos:
    info = np.finfo(tipo)
    dados.append({
        "Tipo": tipo.__name__,
        "Bits": tipo(0).nbytes * 8,
        "Epsilon": info.eps
    })

data_frame = pd.DataFrame(dados)
data_frame
```

## 📈 Visualização gráfica dos valores de epsilon

Para facilitar a comparação entre os tipos numéricos, vou utilizar um gráfico de feito com as bibliotecas `seaborn` e `matplotlib`.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

plt.figure(figsize=(10,6))
sns.lineplot(x="Tipo", y="Epsilon", data=data_frame)
plt.yscale("log")
plt.title("Comparação do Epsilon da Máquina por Tipo de Float")
plt.xlabel("Tipos de float")
plt.ylabel("Epsilon (escala log)")
plt.show()
```

## Precisão com a biblioteca `decimal`

Para contornar problemas que exigem muita precisão como em finanças existem ferramentas como o módulo `decimal` no contexto Python.
Vou reproduzir o algoritmo, mas com `Decimal(1)` ao invés de `float`. Além de aumentar a precisão interna com `getcontext().prec = 60`

```python
from decimal import Decimal, getcontext

# Aqui você pode alterar a precisão 
getcontext().prec = 60

epsilon = Decimal(1)
um = Decimal(1)
soma = um + epsilon
contador = 0

while soma > um:
    epsilon /= 2
    soma = um + epsilon
    contador += 1

print("Precisão mínima (decimal):", epsilon * 2)
print("Iterações com alta precisão:", contador)
```
