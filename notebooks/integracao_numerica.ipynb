# Regra dos Trapézios Repetida

A **Regra dos Trapézios Repetida** é uma das fórmulas de integração numérica baseadas nas fórmulas de Newton-Cotes. Ela consiste em aproximar a integral de uma função por uma soma de áreas de trapézios.

A fórmula geral para a aproximação é:

$$
\int_a^b f(x) \, dx \approx \frac{h}{2} \left[ f(a) + 2 \sum_{i=1}^{n-1} f(a + ih) + f(b) \right]
$$


## Determinação do Número de Sub-Intervalos $n$

Para garantir a precisão desejada, é necessário determinar o número mínimo de sub-intervalos $n$. A fórmula para o erro máximo da regra dos trapézios é dada por:

$$
E_T \leq \frac{(b-a)^3}{12n^2} \cdot \max \left| f''(x) \right|
$$

Onde:

- $a = 0$  
- $b = 1$  
- $n$ = número de sub-intervalos  
- $f''(x)$ = segunda derivada da função original  

Sabemos que:

$$
n = \frac{b-a}{h} = \frac{1}{h}
$$

Aplicando os valores de $a$ e $b$, a expressão para o erro fica:

$$
\frac{1}{12n^2} \cdot \max \left| f''(x) \right| < \frac{1}{6} \times 10^{-4}
$$

## Determinação de $\max \left| f''(x) \right|$

Para encontrar o valor máximo da segunda derivada da função $f(x) = e^{-x^2}$ no intervalo $x \in [0,1]$, podemos derivar a função até obter a segunda derivada:

1. Primeira derivada:

$$
f'(x) = -2e^{-x^2}x
$$

2. Segunda derivada:

$$
f''(x) = -2e^{-x^2} \left(1 - 2x^2\right)
$$
Para encontrar o ponto onde a segunda derivada atinge o valor máximo, pode-se aplicar um novo processo de análise, derivando $f''(x)$ até obter a quarta derivada de $f(x)$, e então igualando a zero:

$$
f^{(4)}(x) = -4e^{-x^2} \left( -4x^4 + 12x^2 - 3 \right) = 0
$$

Resolvendo essa equação, encontramos o único valor de $x$ candidato a ponto crítico dentro do intervalo $[0,1]$ é $\sqrt{\frac{3 - \sqrt{6}}{2}} \approx 0{,}524$. Avaliando nesses pontos e nas extremidades, determinamos:

$$
f''(0) = -2e^{-(0)^2} \left(1 - 2 \cdot (0)^2\right) = -2e^{0} \cdot 1 = -2
$$

$$
f''(1) = -2e^{-(1)^2} \left(1 - 2 \cdot (1)^2\right) = -2e^{-1} \left(1 - 2\right) = \frac{2}{e}
$$

$$
f''(0{,}524) = -2e^{-(0{,}524)^2} \left(1 - 2 \cdot (0{,}524)^2\right) = -0{,}683 
$$

Logo:

$$
\max \left| f''(x) \right| = \frac{2}{e}
$$

Com isso, substituímos na fórmula do erro e resolvemos para $n$:

$$
\frac{n}{12n^3} \cdot \frac{2}{e} < \frac{1}{6} \times 10^{-4}
$$

Simplificando:

$$
\frac{1}{n^2 \cdot e} < 10^{-4}
$$

Multiplicando ambos os lados por $e$:

$$
\frac{1}{n^2} < \frac{10^{-4}}{1/e}
$$

Ou seja:

$$
n^2 \cdot e > 10^4
$$

Então:

$$
n^2 > \frac{10^4}{e}
$$

Finalmente:

$$
n > 10^2 \cdot e^{-1/2} \approx 60,65
$$

Portanto, o valor mínimo de $n$ necessário para atender à precisão solicitada é $n \geq 61$.
Isso garante que o erro estimado da Regra dos Trapézios estará dentro da tolerância especificada.

```python
# Regra dos Trapézios Repetida
def regra_trapezio(x, n):
    somatorio = 0
    for i in range(1, n - 1):
        somatorio += 2 * f(x[i])
    somatorio += f(x[0]) + f(x[-1])
    
    h = (b - a) / (n - 1) # Passo de integração (altura do trapézio)
    integral = (h / 2) * somatorio
    return integral
```

```python
# Número de sub-intervalos
n_t = 61  # Aqui o valor mínimo n_t = 61

# Dividindo o intervalo em n_t pontos
x_t = np.linspace(a, b, n_t)

# Calculando a integral aproximada
integral_t = regra_trapezio(x_t, n_t)

print(f'O valor da integral pela Regra dos Trapézios (n = {n_t}): {integral_t}')
```

# Regra 1/3 de Simpson Repetida

A **Regra 1/3 de Simpson Repetida** é outra fórmula de integração numérica da família de Newton-Cotes, que fornece uma aproximação da integral de uma função através de uma combinação de parábolas.

A fórmula geral da aproximação é:

$$
\int_a^b f(x) \, dx \approx \frac{h}{3} \left[ f(x_0) + 4 \sum_{\text{ímpares}} f(x_i) + 2 \sum_{\text{pares}} f(x_i) + f(x_n) \right]
$$


## Determinação do Número de Sub-Intervalos $n$

Para garantir a precisão desejada, precisamos calcular o número mínimo de sub-intervalos $n$. A fórmula para o erro máximo da Regra 1/3 de Simpson é dada por:

$$
E_S \leq \frac{(b-a)^5}{180n^4} \cdot \max \left| f^{(4)}(x) \right|
$$

Onde:

- $a = 0$  
- $b = 1$  
- $n$ = número de sub-intervalos (deve ser par!)  
- $f^{(4)}(x)$ = quarta derivada da função original  

Sabemos também que:

$$
n = \frac{b-a}{h} = \frac{1}{h}
$$

Aplicando os valores de $a$ e $b$, a expressão para o erro fica:

$$
\frac{1}{180n^4} \cdot \max \left| f^{(4)}(x) \right| < \frac{1}{6} \times 10^{-4}
$$

## Determinação de $\max \left| f^{(4)}(x) \right|$

Para encontrar o valor máximo da quarta derivada da função $f(x) = e^{-x^2}$ no intervalo $x \in [0,1]$, podemos analisar o gráfico de $f^{(4)}(x)$.

Ao observar o gráfico da quarta derivada no intervalo considerado, é possível verificar que o ponto onde a função atinge o valor máximo é em:

$$
x = 0
$$

Neste ponto, temos:

$$
\max \left| f^{(4)}(x) \right| = 12
$$

Esse valor pode ser, facilmente, visualizado pelo gráfico de $f^{(4)}(x)$ a seguir:

```python
import matplotlib.pyplot as plt

# Quarta derivada de f(x)
def d4f(x):
    return -4 * np.exp(-x**2) * (-4 * x**4 + 12 * x**2 - 3)

# Plot do gráfico da quarta derivada de f(x) no intervalo [0, 1]
x_plot = np.linspace(a, b, 1000)
y_plot = d4f(x_plot)

plt.figure(figsize=(8, 4))
plt.plot(x_plot, y_plot, label=r"$f^{(4)}(x)$")
plt.title('Gráfico da Quarta Derivada de $f(x)$ no intervalo [0, 1]')
plt.xlabel('x')
plt.ylabel(r'$f^{(4)}(x)$')
plt.legend()
plt.grid(True)
plt.show()
```

Com o valor de $\max \left| f^{(4)}(x) \right|$ conhecido, podemos substituir na fórmula do erro e resolver para $n$:

$$
\frac{n}{180n^5} \cdot 12 < \frac{1}{6} \times 10^{-4}
$$

Simplificando:

$$
\frac{180n^4}{12} > 6 \times 10^4
$$

$$
n^4 > \frac{72}{180} \times 10^4
$$

$$
n^4 > \frac{2}{5} \times 10^4
$$

$$
n > \sqrt[4]{\frac{2}{5}} \times 10 \approx 7,95
$$

Portanto, o valor mínimo de $n$ necessário para atender à precisão solicitada é, $n \geq 8$.

Lembrando que o $n$ encontrado deve ser um número **par**, devido à condição de aplicação da Regra 1/3 de Simpson.

```python
# Regra de 1/3 de Simpson Repetida
def simpson(x, n):
    somatorio = 0
    for i in range(1, n - 1):
        if i % 2 == 0:
            somatorio += 2 * f(x[i])
        else:
            somatorio += 4 * f(x[i])
    
    somatorio += f(x[0]) + f(x[-1])
    
    h = (b - a) / (n - 1)  # Passo de integração
    integral = (h / 3) * somatorio
    return integral
```

```python
# Número de sub-intervalos (deve ser par)
n_s = 8  # Valor mínimo de n_s calculado anteriormente

# Dividindo o intervalo em n_s pontos
x_s = np.linspace(a, b, n_s)

# Calculando a integral aproximada usando Simpson
integral_s = simpson(x_s, n_s)

print(f'O valor da integral pela Regra de 1/3 de Simpson (n = {n_s}): {integral_s}')
```

## Conclusão Comparativa entre os Métodos Numéricos

Após a aplicação dos dois métodos de integração numérica, os seguintes resultados foram obtidos para o cálculo da integral:

$$
\int_{0}^{1} e^{-x^2} \, dx
$$

### Comparação dos Resultados:

| Método                      | n (Subdivisões) | Resultado Numérico      | Erro Absoluto em relação ao valor de referência  |
|-----------------------------|-----------------|-------------------------|--------------------------------------------------|
| Regra dos Trapézios         | 61              | 0.7468071011991204      | ≈ 1,70 × 10⁻⁵                                    |
| Regra de 1/3 de Simpson     | 8               | 0.7267983960526729      | ≈ 2,00 × 10⁻²                                    |
| Valor de Referência (SciPy) | -               | 0.7468241328124271      | -                                                |

### Análise dos Resultados:

Com base nas deduções teóricas para o erro estimado de cada método, foram escolhidos valores mínimos de \( n \) para tentar atender à precisão desejada.

Observa-se que o **método dos Trapézios**, com $n = 61$ subdivisões, produziu um resultado bastante próximo ao valor de referência, com um erro na ordem de $10^{-5}$. Já o **método de Simpson 1/3**, com apenas $n = 8$ subdivisões, apresentou um resultado razoável, mas com um erro mais significativo, na ordem de $10^{-2}$.

Esses resultados reforçam a teoria de que **métodos de ordem superior**, como o de Simpson, tendem a atingir boa precisão com menos subdivisões, enquanto **métodos mais simples**, como o dos Trapézios, necessitam de um refinamento maior para reduzir o erro.

### Considerações Finais:

É importante destacar que o valor de referência utilizado para comparação foi obtido numericamente por meio do método `quad` da biblioteca **SciPy**, que também é um método numérico, sujeito a um pequeno erro próprio (embora muito inferior ao dos métodos aplicados manualmente).

Além disso, a diferença entre os erros teóricos previstos e os erros reais observados pode ser atribuída a vários fatores, como:

- Estimativas conservadoras feitas para as derivadas das funções nas deduções;
- Diferenças entre o erro teórico de truncamento e o erro real;
- Limitações numéricas e de precisão tanto nas implementações quanto no próprio método de referência.

Essa análise mostra a importância de não confiar apenas em cálculos teóricos, mas também em testes práticos e comparações numéricas reais ao trabalhar com métodos de aproximação.
