## Introdução

Implementação computacional de métodos numéricos para encontrar **zeros de funções reais**, com estudo de caso na função:

$$
f(x) = e^x - 3x
$$

``````python
import numpy as np

# Definindo a função e sua derivada como array NumPy
def f(x):
    return np.exp(x) - 3*x

def df(x):
    return np.exp(x) - 3

#Os retornos de f(x) e df(x) serão "np.float". 
#Por isso em algumas funçoes haverá um cast para float. Especialmente quando for utilizado a função "abs()"
``````

Gráficos de $f(x)$:

``````python
import matplotlib.pyplot as plt

# Criando os pontos para o gráfico
x = np.linspace(-2, 2, 400)
y = f(x)

# Plotando o gráfico
plt.figure(figsize=(8, 5))
plt.axhline(0, color='gray', linestyle='--')  # linha horizontal no y=0
plt.plot(x, y, label=r'$f(x) = e^x - 3x$', color='blue')
plt.title('Gráfico da função $f(x) = e^x - 3x$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
``````

## Estudo dos Sinais da Função

Para determinar os intervalos onde podem existir raízes da função 
$$f(x) = e^x - 3x$$ 
podemos utilizar o Teorema do Valor Intermediário.

Esse teorema nos diz que, se em um intervalo $(a, b)$ temos $f(a) \cdot f(b) < 0$, ou seja, os valores da função em $a$ e $b$ possuem **sinais opostos**, então existe ao menos uma raiz dentro desse intervalo. Isso ocorre porque o gráfico da função deve cruzar o eixo $x$ em algum ponto entre $a$ e $b$.

``````python
import pandas as pd

# Gerando uma lista de valores de x e seus respectivos valores de f(x)
x_values = list(range(-5, 6))
y_values = [round(f(x), 2) for x in x_values]

# Criando o DataFrame
data_frame = pd.DataFrame({'x': x_values, 'f(x)': y_values})

# Exibindo a tabela horizontal
data_frame
``````

Com base nos valores calculados, observamos que ocorrem mudanças de sinal nos intervalos $(0, 1)$ e $(1.1, 2)$. Isso indica a presença de **raízes reais** nesses intervalos, pois a função assume sinais opostos nas extremidades.

Para garantir que **existe apenas uma raiz em cada intervalo**, analisamos o sinal da derivada da função:
$$f'(x) = e^x - 3$$

Note que:
- Em $(0, 1)$, temos $f'(x) < 0$
- Em $(1.1, 2)$, temos $f'(x) > 0$

Como a derivada **não muda de sinal dentro de cada intervalo**, a função é **estritamente monótona** em cada um deles. Portanto, existe **uma única raiz** em cada intervalo identificado.




``````python
# 1° intervalo
a1 = 0
b1 = 1

# 2° intervalo
a2 = 1.1
b2 = 2
``````

## Métodos

### Método da Bissecção

O **Método da Bissecção** é uma técnica numérica usada para encontrar as raízes de uma função contínua. O processo consiste em dividir repetidamente o intervalo em duas partes, buscando reduzir o intervalo onde a raiz pode estar.

O método começa com um intervalo inicial $(a, b)$, onde sabemos que a função muda de sinal, ou seja, $f(a) \cdot f(b) < 0$. A cada iteração, o ponto médio do intervalo é calculado:

$$ \text{midpoint} = \frac{a + b}{2} $$

Se $f(a) \cdot f(\text{midpoint}) < 0$, isso significa que a raiz está entre $a$ e $midpoint$, então o novo intervalo será $(a, midpoint)$. Caso contrário, a raiz estará entre $midpoint$ e $b$, e o intervalo se tornará $(midpoint, b)$.

O processo é repetido até que a diferença entre $a$ e $b$ seja menor que uma tolerância definida, garantindo que o intervalo contenha a raiz com uma precisão desejada.



``````python
def biseccao(start, end, tolerance=tolerance):
  count = 0

  while abs(end - start) > tolerance:
      midpoint = (start + end) / 2
      if f(start) * f(midpoint) < 0:
        end = midpoint
      else:
        start = midpoint

      count += 1

  return {"result" : (start + end) / 2, "iterations" :  count}
``````

``````python
print(f"1° Intervalo (0, 1) {biseccao(start= a1, end= b1, tolerance=tolerance)}")
print(f"2° Intervalo (1.1, 2) {biseccao(start= a2, end= b2, tolerance=tolerance)}")
``````

### Método da Posição Falsa

O **Método da Posição Falsa**, também conhecido como **Regula Falsi**, é um método numérico para encontrar as raízes de uma função contínua. Ele é uma combinação dos métodos da Bissecção e da Secante.

A ideia principal é utilizar dois pontos $a$ e $b$ (onde a função muda de sinal), ou seja, $f(a) \cdot f(b) < 0$ e calcular um ponto $c$, chamado de ponto de posição falsa, usando uma fórmula linear baseada nos valores da função nesses dois pontos.

A fórmula para o ponto de posição falsa é dada por:

$$ c = \frac{a \cdot f(b) - b \cdot f(a)}{f(b) - f(a)} $$

Depois de calcular $c$, verificamos o sinal da função nesse ponto:

- Se $f(a) \cdot f(c) < 0$, isso significa que a raiz está entre $a$ e $c$, então atualizamos o intervalo para $(a, c)$.
- Caso contrário, a raiz está entre $c$ e $b$, e atualizamos o intervalo para $(c, b)$.

O processo continua até que o intervalo seja suficientemente pequeno, ou seja, a diferença entre $a$ e $b$ seja menor que uma tolerância definida.

``````python
def posicao_falsa(start, end, tolerance=tolerance, max_iteration=100):

    for iteration in range(max_iteration):
        c = (start * f(end) - end * f(start)) / (f(end) - f(start))

        if abs(float(f(c))) < tolerance:
            return {"result" : float(c), "iterations" :  iteration}

        if f(start) * f(c) < 0:
            end = c
        else:
            start = c

    print("Número máximo de iterações atingido")
    return {"result" : float(c), "iterations" :  max_iteration}
``````

``````python
print(f"1° Intervalo (0, 1) {posicao_falsa(start= a1, end= b1)}")
print(f"2° Intervalo (1.1, 2) {posicao_falsa(start= a2, end= b2)}")
``````

### Método do Ponto Fixo

O método do ponto fixo consiste em obter uma função de iteração $\phi(x)$ a partir da equação $f(x) = 0$.

Supondo que a função de iteração é **convergente**, utiliza-se uma condição inicial $x_0 \in (a,b)$ e executa-se a rotina:

$$
x_{k+1} = \phi(x_k)
$$

até que $|f(x_k)| < \epsilon$, onde $\epsilon$ é a precisão indicada.

---

#### Construção da função de iteração

A equação dada é:

$$
f(x) = e^x - 3x = 0
$$

Podemos isolar \(x\):

$$
x = \dfrac{e^x}{3}
$$

Logo, a função de iteração é:

$$
\phi_1(x) = \dfrac{e^x}{3}
$$

#### Verificação da convergência de $\phi_1(x)$

A condição de convergência do método do ponto fixo é:

$$
|\phi'(x)| < 1
$$

Calculando a derivada:

$$
\phi_1'(x) = \dfrac{e^x}{3}
$$

Portanto, a condição de convergência se torna:

$$
\dfrac{e^x}{3} < 1 \Rightarrow e^x < 3 \Rightarrow x < \ln(3) \approx 1.0986
$$

**Conclusão:** a função $\phi_1(x) = \dfrac{e^x}{3}$ é **convergente** para:

$$
x < \ln(3) \approx 1.0986
$$

Ou seja:

- O método **converge** no intervalo $(0, 1)$
- O método **não converge** no intervalo $(1.1, 2)$

---

### Análise Alternativa:

Para o segundo intervalo $(1.1, 2)$, consideramos outra função de iteração:

$$
\phi_2(x) = \ln(3x)
$$

Essa função também satisfaz a equação $f(x) = e^x - 3x = 0$, pois:

$$
x = \ln(3x) \Rightarrow e^x = 3x
$$

#### Verificação da convergência de $\phi_2(x)$

A derivada da função é:

$$
\phi_2'(x) = \frac{1}{x}
$$

No intervalo $(1.1, 2)$, temos:

$$
|\phi_2'(x)| = \left| \frac{1}{x} \right| < 1
$$

Logo, a função $\phi_2(x)$ é **contrativa** nesse intervalo, o que garante convergência do método.

Além disso, a imagem da função permanece dentro do próprio intervalo:

$$
\phi_2(1.1) \approx \ln(3.3) \approx 1.194, \quad \phi_2(2) \approx \ln(6) \approx 1.792
$$

Portanto, a função $\phi_2(x) = \ln(3x)$ é **adequada para o segundo intervalo** e permite que o método do ponto fixo **converta corretamente para a segunda raiz**.

Isso mostra como a escolha da função de iteração é fundamental para o sucesso do método: uma função mal escolhida pode levar à divergência ou à convergência para a raiz errada.


``````python
# Funções de iteração
def phi_1(x):
    return np.exp(x) / 3

def phi_2(x):
    return np.log(3 * x)
``````

``````python
def ponto_fixo(phi, x0, tolerance=1e-8, max_iterations=100):
    for i in range(max_iterations):
        x1 = phi(x0)
        if abs(f(x1)) < tolerance or abs(x1 - x0) < tolerance:
            return {"result": float(x1), "iterations": i + 1}
        x0 = x1
    print("Número máximo de iterações atingido")
    return {"result": float(x1), "iterations": max_iterations}
``````

``````python
# Intervalo (0, 1) com phi_1
print("Intervalo (0,1), x0 = 0.5, usando phi_1(x) = e^x / 3")
print(ponto_fixo(phi_1, 0.5))

# Intervalo (1.1, 2) com phi_2
print("\nIntervalo (1.1,2), x0 = 1.5, usando phi_2(x) = ln(3x)")
print(ponto_fixo(phi_2, 1.5))
``````

### Teste Cruzado entre Funções de Iteração e Intervalos

Para reforçar a importância da escolha adequada da função de iteração no método do ponto fixo, realizamos dois testes cruzados, aplicando:

- $\phi_2(x) = \ln(3x)$ no intervalo **(0, 1)**
- $\phi_1(x) = \dfrac{e^x}{3}$ no intervalo **(1.1, 2)**

**Resultado 1:** $\phi_2(x) = \ln(3x)$ no intervalo (0, 1)

A função $\phi_2(x) = \ln(3x)$ requer que $3x > 0$, pois o logaritmo natural não é definido para valores negativos ou nulos. Embora o valor inicial $x_0 = 0.5$ ainda seja válido (pois $3 \cdot 0.5 = 1.5$), a sequência de iterações tende a reduzir os valores de $x$ conforme tenta se aproximar da raiz.

Com isso, eventualmente $3x$ se torna menor que zero, fazendo com que a função $\phi_2$ retorne **valores indefinidos** (`nan`). Isso gera o seguinte aviso durante a execução:

``RuntimeWarning: invalid value encountered in log``


O resultado final é:

```python
{'result': nan, 'iterations': 100}
````

Ou seja, o método falha completamente.

#### Resultado 2: $\phi_1(x) = \dfrac{e^x}{3}$ no intervalo (1.1, 2)

Neste segundo caso, $\phi_1(x)$ está bem definida numericamente, mas **não satisfaz a condição de convergência** no intervalo analisado, pois:

$$
\phi_1'(x) = \dfrac{e^x}{3} > 1 \quad \text{para } x > \ln(3)
$$

Isso significa que $\phi_1(x)$ **não é contrativa**, o que permite que o método "escape" da região da segunda raiz e deslize lentamente até a **primeira raiz**, próxima de 0.619.

O resultado foi:

```python
{'result': 0.6190612998638485, 'iterations': 48}
`````

Ou seja, mesmo com valor inicial em $(1.1, 2)$, o método converge para a raiz errada, confirmando a análise teórica anterior.

``````python
# Intervalo (0, 1) com phi_2
print("Intervalo (0,1), x0 = 0.5, usando phi_1(x) = e^x / 3")
print(ponto_fixo(phi_2, 0.5))

# Intervalo (1.1, 2) com phi_1
print("\nIntervalo (1.1,2), x0 = 1.5, usando phi_2(x) = ln(3x)")
print(ponto_fixo(phi_1, 1.5))
``````

### Método de Newton

O método de Newton é um caso particular do método do ponto fixo, no qual utilizamos a seguinte função de iteração:

$$
\phi(x_k) = x_k - \dfrac{f(x_k)}{f'(x_k)}
$$

Esse método é convergente se:

- $f'(x) \neq 0$ no intervalo analisado;
- $f''(x)$ existir e for contínua em $(a, b)$.

Verificando essas condições para a função escolhida:

- Derivada primeira:
  $$
  f'(x) = e^x - 3 \neq 0
  $$
- Derivada segunda:
  $$
  f''(x) = e^x
  $$
  que é contínua para todo número real.

Portanto, como todas as hipóteses são satisfeitas, o método de Newton pode ser aplicado com segurança à função estudada.

Para garantir uma comparação justa com os demais métodos numéricos, adotamos como valor inicial $x_0$ o ponto médio de cada intervalo:

- **1° intervalo (0, 1)**: $x_0 = 0.5$
- **2° intervalo (1.1, 2)**: $x_0 = 1.55$

``````python
def newton(x0, tolerance=tolerance):
    count = 0
    while abs(float(f(x0))) > tolerance:

        x1 = x0 - f(x0) / df(x0)

        if abs(float(f(x1))) < tolerance or abs(float(x1 - x0)) < tolerance:
            return {"result" : float(x1), "iterations" : count}

        x0 = x1
        count += 1
    return {"result" : float(x0), "iterations" : count}
    
``````

``````python
print(f"1° Intervalo (0, 1) {newton(x0=1)}")
print(f"2° Intervalo (1.1, 2) {newton(x0=2)}")
``````

### Método da Secante

O método da secante é uma discretização do método de Newton, que evita o cálculo direto da derivada $f'(x)$, utilizando em seu lugar uma aproximação baseada em dois pontos consecutivos. Isso o torna especialmente útil quando a derivada da função é difícil ou custosa de obter analiticamente.

Desta forma, as iterações são calculadas por:

$$
x_{k+1} = \dfrac{x_{k-1}\cdot f(x_k) - x_k\cdot f(x_{k-1})}{f(x_k) - f(x_{k-1})}
$$

Ou, de forma equivalente:

$$
x_{k+1} = x_k - f(x_k) \cdot \dfrac{x_k - x_{k-1}}{f(x_k) - f(x_{k-1})}
$$

Assim, são necessárias **duas aproximações iniciais** $x_0$ e $x_1$ dentro do intervalo $(a, b)$ para iniciar o processo iterativo. De maneira análoga ao que foi feito no Método de Newton, utilizaremos $x_0 = a$ e $x_1 = b$

``````python
def secante(x0, x1, tolerance=tolerance, max_iterations=100):
    if abs(float(f(x0))) < tolerance:
        return {"result" : float(x0), "iterations" : 0}
    
    if abs(float(f(x1))) < tolerance or abs(float(x1 - x0)) < tolerance:
        return {"result" : float(x1), "iterations" : 0}

    for iteration in range(1, max_iterations + 1):
        denominator = f(x1) - f(x0)
        if denominator == 0:
            print("Divisão por zero detectada na iteração", iteration)
            return {"result": float(x1), "iterations": iteration}

        x2 = x1 - f(x1) * (x1 - x0) / denominator

        if abs(float(f(x2))) < tolerance or abs(float(x2 - x1)) < tolerance:
            return {"result": float(x2), "iterations": iteration}

        x0, x1 = x1, x2

    print("Número máximo de iterações atingido")
    return {"result": float(x2), "iterations": max_iterations}
``````

``````python
print(f"1° Intervalo (0, 1) {secante(x0= a1, x1= b1)}")
print(f"2° Intervalo (1.1, 2) {secante(x0= a2, x1= b2)}")
``````

O **Método de Newton**, apesar de ser extremamente eficiente, apresenta instabilidade quando a derivada $f'(x)$ se aproxima de zero. Por exemplo, em $x_0 = 1.1$:

$$
f'(1.1) = e^{1.1} - 3 \approx 0.004
$$

Essa pequena derivada torna o denominador da fórmula quase nulo, resultando em **76 iterações** para convergir.

No caso do **método do ponto fixo**, observamos que:

- A função $\phi_1(x) = \frac{e^x}{3}$ é **válida** e convergente no intervalo $(0, 1)$.
- No intervalo $(1.1, 2)$, $\phi_1'(x) > 1$, o que viola a condição de contração, e de fato o método **convergiu para a raiz errada** (a do primeiro intervalo).
- Por outro lado, a função $\phi_2(x) = \ln(3x)$ é **válida e contrativa** no intervalo $(1.1, 2)$, e convergiu corretamente para a segunda raiz.

Esse comportamento reforça que:

- A função $\phi(x)$ **deve ser cuidadosamente escolhida para cada intervalo**.
- O método do ponto fixo é conceitualmente simples, mas exige **análise criteriosa** para garantir convergência.
- Apresenta desempenho inferior aos métodos como Newton ou Secante, devido à sua **convergência linear**, contra a **quadrática** de Newton.

Portanto, embora todos os métodos tenham obtido bons resultados com escolhas adequadas, destaca-se a importância de uma **boa análise teórica prévia**, especialmente no caso do ponto fixo.
