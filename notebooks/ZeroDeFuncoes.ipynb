## Introdução

O objetivo é apresentar a implementação computacional de diferentes métodos numéricos para encontrar os **zeros de funções reais**.

A função utilizada como estudo de caso é:
$$f(x) = e^x - 3x$$


```python
import numpy as np

# Definindo a função e sua derivada como array NumPy
def f(x):
    return np.exp(x) - 3*x

def df(x):
    return np.exp(x) - 3
```

Gráficos de $f(x)$ e $f'(x)$:

```python
import matplotlib.pyplot as plt

# Criando os pontos para o gráfico
x = np.linspace(-2, 2, 400)
y = f(x)

# Plotando o gráfico
plt.figure(figsize=(8, 5))
plt.axhline(0, color='gray', linestyle='--')  # linha horizontal no y=0
plt.plot(x, y, label=r'$f(x) = e^x - 3x$', color='blue')
plt.title('Gráfico da função $f(x) = e^x - 3x$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

## Estudo dos Sinais da Função

Para determinar os intervalos onde podem existir raízes da função 
$$f(x) = e^x - 3x$$ 
podemos utilizar o Teorema do Valor Intermediário.

Esse teorema nos diz que, se em um intervalo $(a, b)$ temos $f(a) \cdot f(b) < 0$, ou seja, os valores da função em $a$ e $b$ possuem **sinais opostos**, então existe ao menos uma raiz dentro desse intervalo. Isso ocorre porque o gráfico da função deve cruzar o eixo $x$ em algum ponto entre $a$ e $b$.

```python
import pandas as pd

# Gerando uma lista de valores de x e seus respectivos valores de f(x)
x_values = list(range(-10, 11))
y_values = [round(f(x), 2) for x in x_values]

# Criando o DataFrame
df = pd.DataFrame({'x': x_values, 'f(x)': y_values})

# Exibindo a tabela horizontal
df
```

Com base nos valores calculados, observamos que ocorrem mudanças de sinal nos intervalos $(0, 1)$ e $(1.1, 2)$. Isso indica a presença de **raízes reais** nesses intervalos, pois a função assume sinais opostos nas extremidades.

Para garantir que **existe apenas uma raiz em cada intervalo**, analisamos o sinal da derivada da função:
$$f'(x) = e^x - 3$$

Note que:
- Em $(0, 1)$, temos $f'(x) < 0$
- Em $(1.1, 2)$, temos $f'(x) > 0$

Como a derivada **não muda de sinal dentro de cada intervalo**, a função é **estritamente monótona** em cada um deles. Portanto, existe **uma única raiz** em cada intervalo identificado.




```python
# 1° intervalo
a1 = 0
b1 = 1

# 2° intervalo
a2 = 1.1
b2 = 2
```

## Métodos

### Método da Bissecção

O **Método da Bissecção** é uma técnica numérica usada para encontrar as raízes de uma função contínua. O processo consiste em dividir repetidamente o intervalo em duas partes, buscando reduzir o intervalo onde a raiz pode estar.

O método começa com um intervalo inicial $(a, b)$, onde sabemos que a função muda de sinal, ou seja, $f(a) \cdot f(b) < 0$. A cada iteração, o ponto médio do intervalo é calculado:

$$ \text{midpoint} = \frac{a + b}{2} $$

Se $f(a) \cdot f(\text{midpoint}) < 0$, isso significa que a raiz está entre $a$ e $midpoint$, então o novo intervalo será $(a, midpoint)$. Caso contrário, a raiz estará entre $midpoint$ e $b$, e o intervalo se tornará $(midpoint, b)$.

O processo é repetido até que a diferença entre $a$ e $b$ seja menor que uma tolerância definida, garantindo que o intervalo contenha a raiz com uma precisão desejada.



```python
def biseccao(start, end, tolerance):
  count = 0

  while end - start > tolerance:
      midpoint = (start + end) / 2
      if f(start) * f(midpoint) < 0:
        end = midpoint
      else:
        start = midpoint

      count += 1

  return {"result" : (start + end) / 2, "iterations" :  count}
```

```python
# Vlor encontrado e número de iteraçoes pelo método de Bisecção
print(f"1° Intervalo (0, 1) {biseccao(start= a1, end= b1, tolerance=tolerance)}")
print(f"2° Intervalo (1.1, 2) {biseccao(start= a2, end= b2, tolerance=tolerance)}")
```

### Método da Posição Falsa

O **Método da Posição Falsa**, também conhecido como **Regula Falsi**, é um método numérico para encontrar as raízes de uma função contínua. Ele é uma combinação dos métodos da Bissecção e da Secante.

A ideia principal é utilizar dois pontos $a$ e $b$ (onde a função muda de sinal), ou seja, $f(a) \cdot f(b) < 0$ e calcular um ponto $c$, chamado de ponto de posição falsa, usando uma fórmula linear baseada nos valores da função nesses dois pontos.

A fórmula para o ponto de posição falsa é dada por:

$$ c = \frac{a \cdot f(b) - b \cdot f(a)}{f(b) - f(a)} $$

Depois de calcular $c$, verificamos o sinal da função nesse ponto:

- Se $f(a) \cdot f(c) < 0$, isso significa que a raiz está entre $a$ e $c$, então atualizamos o intervalo para $(a, c)$.
- Caso contrário, a raiz está entre $c$ e $b$, e atualizamos o intervalo para $(c, b)$.

O processo continua até que o intervalo seja suficientemente pequeno, ou seja, a diferença entre $a$ e $b$ seja menor que uma tolerância definida.
